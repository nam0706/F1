# Kiểm tra max lap cho session_key = 9468
Notebook này đọc file `data/raw/laps.csv`, lọc theo `session_key = 9468`, rồi tính `lap_number` lớn nhất để chứng minh max lap là bao nhiêu.

```python
from pathlib import Path
import pandas as pd

# Đường dẫn từ notebook/nam.ipynb về thư mục gốc project
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebook' else Path.cwd()
LAPS_PATH = PROJECT_ROOT / 'data' / 'raw' / 'laps.csv'

print('File tồn tại:', LAPS_PATH.exists())
print('Đường dẫn:', LAPS_PATH)
```

```python
import pandas as pd
from pathlib import Path

# Định nghĩa đường dẫn folder laps thay vì 1 file đơn lẻ
LAPS_DIR = Path(r'd:\Main class UIT\DS108\F1\data\raw\laps')

def load_laps_data(folder_path):
    if not folder_path.exists():
        print(f"❌ Không tìm thấy folder: {folder_path}")
        return pd.DataFrame()
    
    # Lấy tất cả file .csv và .parquet trong folder
    files = list(folder_path.glob('**/*.csv')) + list(folder_path.glob('**/*.parquet'))
    
    if not files:
        print("❌ Không tìm thấy file dữ liệu nào trong folder laps")
        return pd.DataFrame()
    
    frames = []
    for f in files:
        # Đọc tùy theo đuôi file
        df = pd.read_parquet(f) if f.suffix == '.parquet' else pd.read_csv(f, low_memory=False)
        if not df.empty:
            frames.append(df)
    
    # Gộp tất cả các file lại thành 1 DataFrame
    return pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()

# Thực thi đọc dữ liệu
laps = load_laps_data(LAPS_DIR)

if not laps.empty:
    print('✅ Đã load dữ liệu thành công!')
    print('Số dòng, số cột:', laps.shape)
    print('Các cột:')
    print(laps.columns.tolist())
    display(laps.head())
else:
    print('❌ Lỗi: Không thể load dữ liệu laps.')
```

```python
# Lọc session_key = 9468
SESSION_ID = 9468
session_laps = laps[laps['session_key'] == SESSION_ID].copy()

print(f'Số dòng của session_key={SESSION_ID}:', len(session_laps))
session_laps[['session_key', 'driver_number', 'lap_number', 'lap_duration']].head(10)
```

```python
# Tính max lap_number của toàn bộ session
max_lap = session_laps['lap_number'].max()

print(f'Max lap_number của session_key={SESSION_ID} là:', max_lap)
```

```python
rows_max_lap = laps[laps["lap_number"] == laps["lap_number"].max()]

print("Số dòng đạt max lap_number:", len(rows_max_lap))

rows_max_lap[
    ["session_key", "meeting_key", "lap_number"]
].sort_values(["session_key"])

```

```python
# Chứng minh bằng cách xem max lap theo từng driver
max_lap_by_driver = (
    session_laps
    .groupby('driver_number', as_index=False)['lap_number']
    .max()
    .sort_values('lap_number', ascending=False)
)

max_lap_by_driver
```

```python
# Driver/lap đạt giá trị max_lap
rows_with_max_lap = session_laps[session_laps['lap_number'] == max_lap][
    ['session_key', 'driver_number', 'lap_number', 'lap_duration', 'date_start']
]

rows_with_max_lap
```

```python
SESSION_ID = 9468

# Lọc riêng các dòng thuộc session_key = 9468
session_9468 = laps[laps["session_key"] == SESSION_ID]

# Tìm max lap_number trong session này
max_lap_9468 = session_9468["lap_number"].max()

print("Session key:", SESSION_ID)
print("Số dòng trong session:", len(session_9468))
print("Max lap_number trong session:", max_lap_9468)

# Lấy các dòng đạt max lap_number trong session_key = 9468
rows_max_lap_9468 = session_9468[
    session_9468["lap_number"] == max_lap_9468
]

print("Số dòng đạt max lap_number trong session:", len(rows_max_lap_9468))

rows_max_lap_9468[
    ["session_key", "meeting_key", "driver_number", "lap_number", "lap_duration", "date_start"]
].sort_values(["driver_number"])
```
